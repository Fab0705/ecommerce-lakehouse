```python
from pyspark.sql.functions import col, to_timestamp
```

```python
# 1. Limpieza de Orders
print("Limpiando y validando Orders...")
df_orders_bronze = spark.table("workspace.default.bronze_orders")

df_orders_silver = df_orders_bronze \
    .withColumn("order_purchase_timestamp", to_timestamp(col("order_purchase_timestamp"))) \
    .withColumn("order_approved_at", to_timestamp(col("order_approved_at"))) \
    .withColumn("order_delivered_customer_date", to_timestamp(col("order_delivered_customer_date"))) \
    .withColumn("order_estimated_delivery_date", to_timestamp(col("order_estimated_delivery_date"))) \
    .dropDuplicates(["order_id"])

df_orders_silver.write.format("delta").mode("overwrite").saveAsTable("workspace.default.silver_orders")
```

```python
# 2. Limpieza de Order Items
print("Limpiando y validando Order Items...")
df_items_bronze = spark.table("workspace.default.bronze_order_items")

df_items_silver = df_items_bronze \
    .filter(col("price") > 0) \
    .dropDuplicates(["order_id", "order_item_id"])

df_items_silver.write.format("delta").mode("overwrite").saveAsTable("workspace.default.silver_order_items")
```

```python
# 3. Limpieza de Products
print("Limpiando y validando Products...")
df_products_bronze = spark.table("workspace.default.bronze_products")

df_products_silver = df_products_bronze \
    .fillna({"product_category_name": "unknown"}) \
    .dropDuplicates(["product_id"])

df_products_silver.write.format("delta").mode("overwrite").saveAsTable("workspace.default.silver_products")
```

```python
print("✅ Capa Plata (Limpieza) generada exitosamente. Datos listos para el modelado.")
```

```python
from pyspark.sql.functions import col
```

```python
print("Iniciando integración de fuentes (Silver Layer)...")
```

```python
# 1. Leer las tablas limpias de la capa Silver
df_orders = spark.table("workspace.default.silver_orders")
df_items = spark.table("workspace.default.silver_order_items")
df_products = spark.table("workspace.default.silver_products")
```

```python
# 2. Realizar los Joins para crear una tabla transaccional enriquecida
# Usamos INNER join para ventas (solo items con órdenes válidas) 
# y LEFT join para productos (por si algún producto fue borrado del catálogo)
df_integrated_sales = df_items \
    .join(df_orders, on="order_id", how="inner") \
    .join(df_products, on="product_id", how="left")
```

```python
# 3. Guardar la tabla integrada
df_integrated_sales.write.format("delta").mode("overwrite").saveAsTable("workspace.default.silver_integrated_sales")
```

```python
print("✅ Capa Plata (Integración) generada exitosamente. Dataset enriquecido listo para la capa Oro.")
```

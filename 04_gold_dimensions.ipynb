```python
from pyspark.sql.functions import col, monotonically_increasing_id, to_date, year, month, dayofmonth
```

```python
# 1. Crear Dimensión Producto (dim_product)
df_products = spark.table("workspace.default.silver_products")

dim_product = df_products.select(
    col("product_id").alias("product_business_key"),
    col("product_category_name").alias("category"),
    col("product_weight_g").alias("weight_g")
).withColumn("product_key", monotonically_increasing_id()) # Llave subrogada

dim_product.write.format("delta").mode("overwrite").saveAsTable("workspace.default.gold_dim_product")
print("- dim_product generada.")
```

```python
# 2. Crear Dimensión Fecha (dim_date) a partir de las órdenes
df_orders = spark.table("workspace.default.silver_orders")

dim_date = df_orders.select(to_date(col("order_purchase_timestamp")).alias("date_key")).distinct().dropna()

dim_date = dim_date.select(
    col("date_key"),
    year("date_key").alias("year"),
    month("date_key").alias("month"),
    dayofmonth("date_key").alias("day")
)

dim_date.write.format("delta").mode("overwrite").saveAsTable("workspace.default.gold_dim_date")
print("- dim_date generada.")
```

```python
print("✅ Capa Oro (Dimensiones) generada exitosamente.")
```

```python
# Ruta base del volumen en Unity Catalog donde subiste los CSV
volume_path = "/Volumes/workspace/default/olist_raw_data/"
```

```python
# 1. Ingesta de Orders
print("Procesando Orders...")
df_orders = spark.read \
    .format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load(f"{volume_path}olist_orders_dataset.csv")

df_orders.write.format("delta").mode("overwrite").saveAsTable("workspace.default.bronze_orders")
```

```python
# 2. Ingesta de Order Items
print("Procesando Order Items...")
df_order_items = spark.read \
    .format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load(f"{volume_path}olist_order_items_dataset.csv")

df_order_items.write.format("delta").mode("overwrite").saveAsTable("workspace.default.bronze_order_items")
```

```python
# 3. Ingesta de Products
print("Procesando Products...")
df_products = spark.read \
    .format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load(f"{volume_path}olist_products_dataset.csv")

df_products.write.format("delta").mode("overwrite").saveAsTable("workspace.default.bronze_products")
```

```python
print("✅ Capa Bronce generada exitosamente. Datos crudos guardados como tablas Delta.")
```

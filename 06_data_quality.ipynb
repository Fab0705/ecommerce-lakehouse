```python
from pyspark.sql.functions import col
```

```python
print("-" * 30)
print(" 📊 DATA QUALITY REPORT ")
print("-" * 30)

# Cargar la tabla principal de hechos
df_sales = spark.table("workspace.default.gold_fact_sales")

# 1. Conteo total de registros procesados
total_records = df_sales.count()

# 2. Detección de nulos en llaves foráneas críticas (Integridad Referencial)
null_products = df_sales.filter(col("product_key").isNull()).count()
null_dates = df_sales.filter(col("date_key").isNull()).count()

# 3. Validación de Reglas de Negocio (Precios o envíos negativos)
invalid_prices = df_sales.filter((col("price") < 0) | (col("freight_value") < 0)).count()

# 4. Cálculo de fiabilidad (%)
total_errors = null_products + null_dates + invalid_prices
valid_records_pct = ((total_records - total_errors) / total_records) * 100 if total_records > 0 else 0

# Imprimir el reporte ejecutivo
print(f"Registros procesados (Capa Oro): {total_records:,}")
print(f"Llaves de producto nulas:        {null_products}")
print(f"Llaves de fecha nulas:           {null_dates}")
print(f"Precios/Envíos inválidos (< 0):  {invalid_prices}")
print("-" * 30)
print(f"✅ Tasa de registros válidos:    {valid_records_pct:.2f}%")
print("-" * 30)
```
